```python
import pandas as pd

data = {
    'name': ['김철수', '이영희', '박민수', '최지훈', '정소희'],
    'grade': [1, 2, 3, 4, 2],
    'gpa': [4.2, 3.8, 4.5, 3.9, 3.5],
    'department': ['컴퓨터공학', '경영학', '전자공학', '의학', '심리학'],
    'circle': ['프로그래밍', '독서토론', '로봇공학', '봉사활동', '음악감상']
}

# DataFrame 인덱스 지정
df = pd.DataFrame(data)
df = df.set_index('name')
df

# DataFramee row 선택
# Index로 지정된 column의 값만 선택 가능
df.loc['김철수']
df.loc[['김철수', '이영희']]
df.loc['김철수':'최지훈']

# DataFrame 위치 기반 row 선택
df.iloc[1]
df.iloc[[2, 3]]
df.iloc[1:3]

# DataFrame 조건 기반 선택
df.loc[[True, False, True, False, False]]
print(df['grade'] >= 4.0)
df.loc[df['grade'] >= 4.0]
df[df['grade'] == 2]
df[df['grade'] > 2]
df[(df['grade'] == 2) & (df['gpa'] >= 3.7)]
```

```python
import pandas as pd

data = {
    'name': ['김철수', '이영희', '박민수', '최지훈', '정소희'],
    'grade': [1, 2, 3, 4, 2],
    'gpa': [4.2, 3.8, 4.5, 3.9, 3.5],
    'department': ['컴퓨터공학', '경영학', '전자공학', '의학', '심리학'],
    'circle': ['프로그래밍', '독서토론', '로봇공학', '봉사활동', '음악감상']
}

# DataFrame 인덱스 지정
df = pd.DataFrame(data)
df = df.set_index('name')
df

# DataFrame 단일 column 오름차순, 내림차순 정렬
df.sort_values(by='gpa', ascending=False)

# DataFrame 복수 column 오름차순, 내림차순 정렬
df.sort_values(by=['grade', 'gpa'], ascending=[True, False])

# DataFrame 인덱스 기본 오름차순, 내림차순 정렬
df.sort_index(ascending=False)
```
